```python
import velot
import scanpy as sc

import matplotlib.pyplot as plt
import matplotlib.patches as patches
```

```python
adata = sc.read_h5ad("/media/user/T7 Touch/PhD_Lucas/velot_peer_review/velot/notebooks/revision/benchmark/results2/data/velot_seed0_smooth_pancreas.h5ad")
adata
```

## Trajectories for cluster Alpha

```python
start_cluster = "Alpha"
n_trajectories = 15

velot.tl.compute_trajectories(
    adata,
    basis=f"X_pca",
    velocity_key=f"velot_velocity_pca",
    direction="backward",
    use_network=False,
    evolve_pseudotime=True,
    start_cluster=start_cluster,
    n_trajectories=n_trajectories,
    cluster_key="clusters",
    n_steps=300, step_size=2
)
```

```python
n_cols = 5
n_rows = (n_trajectories + n_cols - 1) // n_cols
fig, axes = plt.subplots(n_rows, n_cols, figsize=(4*n_cols, 4*n_rows))
axes = axes.flatten()
for i in range(n_trajectories):
    velot.pl.trajectories(
        adata, color="clusters", trajectory_ids=[i], line_width=1.5, line_style="-", arrow_frequency=5, arrow_size=15,
        basis="umap", show=False, save=False, figsize=(5,5), ax = axes[i], title=i
    )
axes[5].add_patch(
    patches.Rectangle(
        (0, 0), 1, 1,            # Use (-0.03, -0.03), 1.06, 1.06 for outer padding
        transform=axes[5].transAxes,
        fill=False,
        edgecolor="red",
        linewidth=3,
        clip_on=False,           # Prevents the border from being clipped at the edges
        zorder=101
    )
)
for ax in axes[n_trajectories:]:
    ax.remove()
plt.tight_layout()
plt.show()

velot.pl.trajectories(
    adata, color="clusters", trajectory_ids=[5], line_width=1.5, line_style="-", arrow_frequency=5, arrow_size=15,
    basis="umap", show=False, save=True, save_path=f"figure2_j_{start_cluster.lower()}.png"
)
```
